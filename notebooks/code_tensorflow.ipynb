```python
import tensorflow as tf 
from tensorflow.keras.layers import Dense # Module de création de couche
from tensorflow.keras import Sequential # Module pour créér des couche de bout en bout
```

#### Création du modèle séquentiel

```python
model = Sequential([
    Dense(5, input_shape=(2,)),
    Dense(4, activation='relu'),                   
    Dense(2, activation='softmax')                                      
])
#Subclassing
```

$x=[x_1,x_2]^T \in \mathbb{R}^{1 \times 2}$

Calcul de la première couche:  $Z_1= (x^T \times M_1) \in \mathbb{R}^{1 \times 5}  \quad |  \quad M_1 \in \mathbb{R}^{2 \times 5}$ 

Calcul de la deuxième couche:  $Z_2= Relu(Z_1 \times M_2) \in \mathbb{R}_{>0}^{1 \times 4}  \quad | \quad M_2 \in \mathbb{R}^{5 \times 4}$

Calcul de la dernière couche:  $Z_3= Softmax(Z_2 \times M_3) \in \mathbb{R}_{]0,1[}^{1 \times 2}  \quad | \quad M_3 \in \mathbb{R}^{4 \times 2}$

#### Exemple

```python
x = tf.random.normal(shape=(1, 2))
print(f'Dimenssion de x {x.shape}')

#Passons x dans notre modèle
prediction=model(x)
print(prediction)
```

##### Fonctions de perte

```python
a_predire= tf.convert_to_tensor([1.,0])
print(f"Dimenssion de vecteur à prédire {a_predire.shape}")

#Cross entropy loss
cross_entropy = tf.keras.losses.CategoricalCrossentropy()
ce_loss = cross_entropy(a_predire, tf.reshape(prediction, [-1]))
print(f"Cross entropy loss={ce_loss:.2f}")

# Mean Squared Error loss
# mse = tf.keras.losses.MeanSquaredError()
# mse_loss = mse(a_predire, prediction)
```

#### Calcul de gradients

$y=ax+b$

$z=y^2$

$\frac{\partial z}{\partial a}=\frac{\partial z}{\partial y}\frac{\partial y}{\partial a}=(2y)(x)$

$\frac{\partial z}{\partial b}=\frac{\partial z}{\partial y}\frac{\partial y}{\partial b}=(2y)(1)$

```python
x=tf.constant(2.0) #Input
a=tf.Variable(3.)# 3.0==3. ; Paramètre
b=tf.Variable(0.5) #Paramètre

with tf.GradientTape() as tape:
    y=a*x+b
    print(f"y={y}")
    z=y**2
    print(f"z={z}")

dz_da,dz_db=tape.gradient(z,[a,b])

print(f"dz/da={dz_da}") # (2*6.5)(2)
print(f"dz/db={dz_db}") # (2*6.5)(1)
```
