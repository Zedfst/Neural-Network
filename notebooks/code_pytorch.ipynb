```python
import torch
import torch.nn as nn
```

#### Création du modèle séquentiel

```python
model = nn.Sequential(
    #Première couche
    nn.Linear(2, 5),         
    #Deuxième couche
    nn.Linear(5, 4),          
    nn.ReLU(),
    #Dernière couche
    nn.Linear(4, 2),         
    # nn.Softmax(dim=-1)        
)
```

$x=[x_1,x_2]^T \in \mathbb{R}^{1 \times 2}$

Calcul de la première couche:  $Z_1= (x^T \times M_1) \in \mathbb{R}^{1 \times 5}  \quad |  \quad M_1 \in \mathbb{R}^{2 \times 5}$ 

Calcul de la deuxième couche:  $Z_2= Relu(Z_1 \times M_2) \in \mathbb{R}_{>0}^{1 \times 4}  \quad | \quad M_2 \in \mathbb{R}^{5 \times 4}$

Calcul de la dernière couche:  $Z_3= Softmax(Z_2 \times M_3) \in \mathbb{R}_{]0,1[}^{1 \times 2}  \quad | \quad M_3 \in \mathbb{R}^{4 \times 2}$

#### Exemple

```python
x = torch.randn(1, 2)
print(f'Dimenssion de x {x.shape}')

#Passons x dans notre modèle
prediction=model(x)
print(prediction)
```

##### Fonctions de perte

```python
a_predire= torch.tensor([1.,0])
print(f"Dimenssion de vecteur à prédire {a_predire.size()}")
 
cross_entropy= nn.CrossEntropyLoss()
ce_loss=cross_entropy(torch.squeeze(prediction),a_predire)
print(f"Cross entropy loss={ce_loss:.2f}")

# Mean Squared Error loss
# mse = nn.MSELoss()
# mse_loss = mse(prediction, a_predire)
```

#### Calcul de gradients

$y=ax+b$

$z=y^2$

$\frac{\partial z}{\partial a}=\frac{\partial z}{\partial y}\frac{\partial y}{\partial a}=(2y)(x)$

$\frac{\partial z}{\partial b}=\frac{\partial z}{\partial y}\frac{\partial y}{\partial b}=(2y)(1)$

```python
x=torch.tensor(2.0) #Input
a=torch.tensor(3.,requires_grad=True)# 3.0==3. ; Paramètre
b=torch.tensor(0.5,requires_grad=True) #Paramètre

#y=ax+b 
y=a*x +b
print(f"y={y}")
z=y**2
print(f"z={z}")
z.backward()

print(f"dz/da={a.grad}") # (2*6.5)(2)
print(f"dz/db={b.grad}") # (2*6.5)(1)
```
